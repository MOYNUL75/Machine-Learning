```python
#Max Value using For loop
number = [1, 2, 3, 4, 5, 6]
max_value = number[0]

for i in number:
    if(i>max_value):
        max_value = i
print(max_value)
```

```python
#Min Value using For loop
number = [1, 2, 3, 4, 5, 6, -9]
min_value = number[0]

for i in number:
    if(i<min_value):
        min_value = i
print(min_value)
```

```python
#Looping array index
number = [1, 2, 3, 4, 5, 6, 0]

for num in number:
    print(num)
```

```python
#Adding Array Elements
number = [1, 2, 3, 4, 5, 6, 0]
print("Without add new value : ", number)

number.append(75)
print("With add new value    : ", number)
```

```python
#Removing Array Elements
number = [1, 2, 3, 4, 5, 6, 0]
print("Without remove any value  : ", number)

number.pop(5) #Index number
print("With remove 5 index value : ", number)

number.remove(4) #value not index
print("With remove 4  value :  ", number)
```
