```python
# 1+3+5+.....+100=?
sum = 0
for i in range(1, 101):
  if i%2==0:
    continue
  else:
    sum += i #sum = sum + i
print(sum)
```

```python
# 1+3+5+.....+100=?
sum = 0
for i in range(1, 101):
  if i%2!=0:
    sum += i #sum = sum + i
print(sum)
```

```python
# 1+3+5+.....+100=?
sum = 0
for i in range(1, 101, 2):
    sum += i #sum = sum + i
print(sum)
```

While

```python
#syntax
#while condition:
#  statement

i = 1
sum = 0
while i < 101:
  sum = sum + i
  i = i + 1
print(sum)
```

```python
i = 1
sum = 0
while i < 101:
  sum = sum + i
  i = i + 2
print(sum)
```
